```python
import os, sys
from pathlib import Path
import pandas as pd
ROOT = Path(os.environ.get("REPO_ROOT", "/kaggle/working/ML-for-Wireless-Security-Threats-Identification"))
sys.path.insert(0, str(ROOT))
from shared.data_loader import split_all
SOURCE = Path(os.environ.get("AWID3_CLEAN", "/kaggle/input/awid3-clean/awid3_clean.parquet"))
OUTPUT_DIRECTORY = Path("/kaggle/working")
```

```python
dataframe = pd.read_parquet(SOURCE)
train, validation, test = split_all(dataframe)
for name, part in (("train", train), ("val", validation), ("test", test)):
    part.to_parquet(OUTPUT_DIRECTORY / f"{name}.parquet", compression="zstd", index=False)
    print(name, len(part), part["Label"].value_counts().to_dict())
```
